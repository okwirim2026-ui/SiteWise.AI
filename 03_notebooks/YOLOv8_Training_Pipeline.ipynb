## Cell 1 — Provision Environment & GPU Acceleration

```python
!pip install -q ultralytics pyyaml requests
import os
import torch
import ultralytics
ultralytics.checks()
print(f"CUDA available: {torch.cuda.is_available()}")
```

## Cell 2 — Stream Dataset Package from Cloud Base

```python
!wget -q https://githubusercontent.com
!python download_dataset.py
yaml_path = "./production_yolo_dataset/dataset.yaml"
print("Master configuration file path locked at:", os.path.abspath(yaml_path))
```

## Cell 3 — Train the Vision Model (1024x1024 Resolution)

```python
from ultralytics import YOLO
model = YOLO('yolov8s.pt')
results = model.train(
    data=yaml_path,
    epochs=100,
    imgsz=1024,
    batch=16,
    project='sitewise_mep_vision',
    name='run_1',
    seed=0,
    device=0
)
```

## Cell 4 — Evaluate Real Domain Metrics & Save Checkpoint

```python
import shutil
val_metrics = model.val(data=yaml_path, split='val')
print(f"Validation mAP50 Accuracy: {val_metrics.box.map50:.4f}")
run_dir = str(results.save_dir)
shutil.copy(f"{run_dir}/weights/best.pt", "/content/best.pt")
print("SUCCESS: Model weights optimized! Right-click 'best.pt' in the left file pane to download.")
```
